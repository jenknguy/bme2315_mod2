The problem:
Facioscapulohumeral muscular dystrophy (FSHD) progresses slowly and sporadically. A consistent measurement of disease progression is difficult to obtain. There is no approved therapy for the disease, and clinical trials need reliable data that can detect change over relatively short periods of time. One approach is to use candidate biomarkers, such as the MRI fat fraction of the tibialis anterior (TA). A prior study found that whole-muscle fat fraction was more strongly associated with the DUX4 molecular signature measured in muscle biopsies than fat fraction in a single region of the muscle (Wong et al., 2024), but it did not compare how well the measures detect change over time. The alternative options are functional and clinical measures, such as the Timed Up and Go (TUG), 10-meter walk/run, stair ascent and descent, dorsiflexion strength, and the Clinical Severity Score (CSS). Using data from 34 FSHD subjects, we evaluate which candidate measure is most sensitive to change and should be recommended as an exploratory endpoint.

The question:
In this project, our team seeks to compare MRI fat fraction data to functional measures. Is whole-muscle fat fraction or a functional or clinical measure, such as TUG, 10-m walk/run, stair ascent/descent, dorsiflexion strength, or CSS, more sensitive to change over time? We define sensitivity as the standardized response mean (the mean change from baseline divided by the standard deviation of that change), calculated for each measure.

Why it matters:
When conducting a trial, researchers need results that are reliable, consistent, and accurate. Clunky measurements can slow down a trial and cause more hardship for the patients and researchers conducting the study. In addition, measures that avoid invasive testing, such as muscle biopsy, reduce the burden on trial participants. Measures that give results that are hard to interpret can delay much-needed research findings. The type of test also affects who can be included in a trial, because MRI is costly and less accessible than a timed walk. On the other hand, a functional test can miss early muscle change.

## Disease Background:

* Prevalence & incidence
Novartis estimates that FSHD affects roughly 45,000–87,000 people in the US and EU. A study in the Netherlands estimated that about 52 people are newly diagnosed each year on average, an incidence of 0.3 per 100,000 person-years, with a prevalence of 12 per 100,000.
(https://www.novartis.com/news/media-releases/novartis-delpacibart-braxlosiran-del-brax-phase-iii-study-facioscapulohumeral-muscular-dystrophy-fshd-meets-primary-biomarker-endpoint, https://pubmed.ncbi.nlm.nih.gov/25122204/)

* Symptoms
FSHD is characterized by progressive muscle degeneration, and symptoms can take a long time to develop. Most patients are not diagnosed until their 20s, and some do not notice signs until their 50s. Common symptoms include facial weakness, shoulder weakness (scapular winging), lower leg weakness (foot drop), abdominal weakness and posture changes, hip (pelvic girdle) weakness, and pain and inflammation.
(http://mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms)

* Diagnosis
Diagnosis starts with a doctor's visit and a physical exam. An initial diagnosis is based on the pattern of muscles affected, along with the patient's family and medical history. The doctor may then order tests, such as blood tests, neurological tests, or muscle biopsies, to find out whether FSHD is causing the problems.
(https://www.fshdsociety.org/living-with-fshd/understanding-fshd/diagnosis/)

* Standard of care treatments (& reimbursement)
There is currently no cure and no disease-specific treatment for FSHD. Care focuses on managing symptoms, maintaining function, and supporting patients' quality of life. It can include pain and symptom management, rehabilitation and supportive care, surgery, exercise, and nutrition.
(https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/medical-management)

* Disease progression & prognosis
The severity and progression of FSHD vary widely, but the disease usually progresses slowly. Most people have a normal life expectancy, although in some cases, especially more severe forms, respiratory problems can occur and may be serious.
(https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy)

* Biological mechanisms
FSHD results from inefficient silencing of the D4Z4 repeat region on chromosome 4. As a result, the DUX4 gene, which codes for a transcription factor, is switched on unevenly in skeletal muscle. DUX4 is normally active in the testis and silenced in other tissues. When it is active in muscle, it turns on many genes normally active in germ cells and stem cells, among others, which may explain how FSHD develops.
(https://pmc.ncbi.nlm.nih.gov/articles/PMC4060068/)

* Clinical trials / next-gen therapies
Multiple trials are currently underway. Examples include:
- Phase 3: Novartis FORTITUDE-3, an RNA therapy that blocks DUX4.
- Phase 2: Scholar Rock FORGE, an antibody treatment that blocks myostatin.
- Phase 1/2: Novartis FORTITUDE-1/2 (enrollment complete), an RNA therapy that blocks DUX4; Sarepta SRP-1001 (formerly ARO-DUX4), an RNA therapy that blocks DUX4.
- Phase 1: KUMC Target FSHD, a beta-agonist that blocks DUX4; RESTEM Restem-L, testing the safety and early effects of a stem cell product derived from umbilical cord lining (ULSC).
- Observational studies: KUMC MOVE and MOVE+, a natural history study to help improve future trials; KUMC MOVE Peds, a natural history study for children with FSHD.
(https://www.fshdsociety.org/research-and-clinical-care/current-trials-and-studies/)

## Data Set:


- The subject and visit/timepoint(s) analyzed
This project uses two data sources.

1. Clinical dataset. In the clinical dataset, the study cohort included 34 FSHD subjects, 16 female and 18 male, aged 21 to 69. Participants came from three sites: the University of Washington Medical Center, the University of Rochester Medical Center, and the Kansas University Medical Center (Wong et al., 2024).
Each participant underwent initial functional testing, which included lower extremity MRI, lower extremity dorsiflexor strength measurements, and needle muscle biopsy of both tibialis anterior (TA) muscles. The biopsies were usually done within a few days of the MRI and strength testing. Of the 68 TA muscles, 43 were STIR− (rating 0) and 25 were STIR+ (rating 1–4) (Wong et al., 2024).
The CSV has baseline, y1, and y2 visits, with timing in lag_months (y1 median 12.4 months, y2 around 25 months) (Module2_FSHD_Dataset.csv).

2. MRI images (FSHD_MRI_data). These are images of the left TA from one participant at two visits (TP1 and TP2) (FSHD_MRI_data README.md and dataset_info.json).

- The tibialis anterior MRI data
All MRI examinations were performed on 3T Siemens scanners running E11C. T1 and T2 Dixon and STIR sequences were centered around the tibial spine and obtained with flexible array coils. Fiducial markers of the mid-portion of the TA muscle were used to verify the muscle biopsy site; however, the muscle STIR score was based on the STIR signal for the whole muscle (Wong et al., 2024).

- The water-image and fat-image files
Fat infiltration percentage was calculated for each pixel as the pixel's fat series intensity divided by the sum of the pixel's fat series intensity and water series intensity (Wong et al., 2024; Ma, 2008). Whole TA fat infiltration was found by taking the mean fat infiltration across the pixels corresponding to the labeled TA (Wong et al., 2024).
Each of the slices has one water file and one fat file, each are 266 × 224, with 1.116 × 1.116 mm in-plane resolution and 4.0 mm slice thickness. The arrays contain the TA-containing half of each image (FSHD_MRI_data README.md and dataset_info.json).

- The tibialis anterior segmentation mask
The TA was segmented by an AI-based algorithm (Wong et al., 2024; Ni et al., 2019). The AI segmentation was verified by a team of trained engineers and evaluated by a single reviewer to ensure consistency (Wong et al., 2024).

- The sampled locations along the muscle
A vectorized version of fat infiltration along the length of the muscle was found by computing fat infiltration of the TA at each axial slice as a function of axial slice progression from distal to proximal. Figure 1b of the paper shows fat infiltration percent over the length of each TA. The biopsy regional sample was a small region of interest placed using fiducial landmarks at the mid-portion of the TA (Wong et al., 2024).

- The meaning and units of the image signals
Fat infiltration is a ratio of fat signal to the sum of fat and water signal. This is reported as a percentage in the study (Wong et al., 2024).

- The clinical, biomarker, or functional measurements included in the dataset
TA strength was measured in kilograms using a fixed system with a force transducer. In the 68 TA muscles, it ranged from 1.8 to 43 kg. Patients were examined and given a 10-point Clinical Severity Score. The DUX4 score was derived from the sum of log10(TPM + 1) over a six-gene DUX4 basket, from RNA sequencing of the biopsies. Lastly, STIR hyperintensity was rated qualitatively on a four-point scale, scored first by two raters, with discordant ratings resolved by consensus (Wong et al., 2024).

## MRI Data Analysis Methodology

To analyze the MRI data, Python was used to process water, fat, and segmentation-mask images of the tibialis anterior (TA) muscle in a patient with facioscapulohumeral muscular dystrophy (FSHD). The NumPy library was used to load and process the MRI images, Pandas was used to organize the data, and Matplotlib was used to generate visualizations. The dataset contained eight cross-sectional MRI slices collected at different locations along the tibialis anterior muscle at two timepoints, TP1 and TP2. The water, fat, and segmentation-mask images were loaded using the np.load() function, while the slice_info.csv file was used to identify the corresponding images and their locations along the muscle. The segmentation mask was converted into a Boolean array to identify pixels belonging specifically to the tibialis anterior muscle. This allowed the muscle to be isolated from surrounding tissues, preventing fat signals outside the muscle from influencing the muscle-specific measurements. 

The fat fraction was calculated for each pixel by dividing the fat signal intensity by the sum of the water and fat signal intensities and multiplying by 100 to obtain a percentage, using the equation: 

Fat Fraction (%) = [Fat Signal / (Water Signal + Fat Signal)] × 100 

To prevent division by zero, the calculation was performed only for pixels where the combined water and fat signal was greater than zero. The fat fraction was first calculated across the whole supplied image region and then restricted to the tibialis anterior muscle using the segmentation mask. Pixels outside the muscle were assigned NaN values to exclude them from the muscle-specific statistical analysis. To visualize differences in fat distribution, the water and fat images were displayed alongside the segmentation mask and calculated fat-fraction maps. The segmentation mask was outlined on the MRI images to identify the boundaries of the tibialis anterior muscle. A separate fat-fraction map was generated for the whole image region and the isolated TA muscle, allowing differences in fat accumulation to be visualized. Histograms were also generated to compare the distribution of fat-fraction values between the whole image region and the TA muscle. The x-axis represented fat fraction as a percentage, while the y-axis represented the number of pixels corresponding to each fat-fraction range. This comparison demonstrated the importance of segmentation in obtaining measurements that specifically represent muscle tissue rather than surrounding structures. 

To quantitatively analyze the MRI data, the mean, median, standard deviation, and number of muscle pixels were calculated for each MRI slice. The mean represented the average fat fraction within the muscle, while the median represented the middle value of the fat-fraction distribution. The standard deviation was used to characterize variability in fat fraction across the muscle pixels, and the number of muscle pixels was calculated from the segmentation mask. The analysis was repeated across eight sampled locations along the tibialis anterior muscle. For each location, the mean fat fraction was calculated and recorded alongside the physical distance from the beginning of the muscle. This allowed regional differences in fat accumulation to be examined rather than relying on a single cross-sectional measurement. 

To examine changes in muscle composition over time, the mean fat fraction was compared between TP1 and TP2. The same fat-fraction calculation and segmentation methodology were applied to both visits. Mean fat fraction was plotted against physical distance along the muscle to visualize differences between the two timepoints. The analysis demonstrated regional differences in fat fraction, with higher fat-fraction values observed at most sampled locations during TP2 compared with TP1. These findings suggest an increase in fat accumulation within the sampled tibialis anterior muscle regions over time. However, changes in MRI fat fraction do not necessarily indicate corresponding changes in muscle strength or physical function. Overall, this methodology demonstrates how MRI fat-fraction measurements can be generated and used to characterize changes in muscle composition. These measurements may serve as potential biomarkers for monitoring FSHD progression and can be compared with functional and clinical outcomes to determine their usefulness as exploratory endpoints in clinical trials.

def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
