* Prevalence & incidence
FSHD affects approximately 45-87k people in the US and EU.
One study conducted in the netherlands found that on average, 52 people were newly diagnosed with FSHD every year. This results in an incidence rate of 0.3/100,000 person-years in the Netherlands. The prevalence rate was 12/100,000.
(https://www.novartis.com/news/media-releases/novartis-delpacibart-braxlosiran-del-brax-phase-iii-study-facioscapulohumeral-muscular-dystrophy-fshd-meets-primary-biomarker-endpoint, and
https://pubmed.ncbi.nlm.nih.gov/25122204/)
* Symptoms
Symptoms may take a very long time to develop.  Most patients are not diagnosed with the disease until their 20s, and some may not notice indications until their 50s. It is characterized by progressive muscle degeneration. Common symptoms are facial weakness, Shoulder weakness (scapular winging), Lower leg weakness (foot drop), Abdominal weakness and posture changes, Hip (pelvic girdle) weakness, and Pain and inflammation.
(http://mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms)
* Diagnosis
 The first step in diagnosing FSHD is to visit with a doctor and get a physical exam. An initial diagnosis is based on the pattern of muscles affected. The doctor will ask questions about the patient’s family history and medical history.
 Then, the doctor may order tests to determine whether the problems are a result of FSHD. These tests can include blood tests, Neurological tests, and Muscle biopsies.
 (https://www.fshdsociety.org/living-with-fshd/understanding-fshd/diagnosis/)
* Standard of care treatments (& reimbursement)
There is currently no cure or disease-specific treatment for FSHD. Care prioritizes managing symptoms, maintaining function, and supporting the patients quality of life.
This care can include pain and symptom management, rehabilitation and supportive care, surgery, exercise, and nutrition.
(https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/medical-management)
* Disease progression & prognosis
The severity and progression of FSHD vary widely. IT usually progresses slowly. Most people have a normal life expectancy. However, in some cases, especially more severe forms, respiratory issues can occur and may be serious.
(https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy)
* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
FSHD is caused by inefficient repeat-mediated epigenetic repression of the D4Z4 macrosatellite repeat array on chromosome 4, resulting in the variegated expression of the DUX4 retrogene, encoding a double-homeobox transcription factor, in skeletal muscle. Normally expressed in the testis and epigenetically repressed in somatic tissues, DUX4 expression in skeletal muscle induces expression of many germline, stem cell, and other genes that might account for the pathophysiology of FSHD. 
(https://pmc.ncbi.nlm.nih.gov/articles/PMC4060068/)
* Clinical Trials/next-gen therapies
There are multiple trials currently underway. A few in different phases are:
Phase 3:
Novartis FORTITUDE-3 – Phase 3 study of an RNA therapy that blocks DUX4
Phase 2:
Scholar Rock FORGE – Phase 2 study of an antibody treatment that blocks myostatin
Novartis FORTITUDE-1/2 (Enrollment complete) – Phase 1/2 study of an RNA therapy that blocks DUX4
Phase 1:
KUMC Target FSHD – Phase 1 study of a Beta-agonist that blocks DUX4
RESTEM Restem-L – Testing the safety and early effects of umbilical cord lining-derived stem cell product (ULSC)
Observational Studies:
KUMC MOVE and MOVE+ – Natural history study to help improve future clinical trials
KUMC MOVE Peds – Natural history study for children with FSHD
Phase 1/2:
Sarepta SRP-1001 (formerly ARO-DUX4) – Phase 1/2 study of an RNA therapy that blocks DUX4
(https://www.fshdsociety.org/research-and-clinical-care/current-trials-and-studies/)
